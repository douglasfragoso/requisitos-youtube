# 04 — Evidências de requisitos

Liga sentenças ao STM documental e cria uma tabela de revisão humana. Não aceita requisitos automaticamente.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().resolve().parents[1]
sys.path.insert(0, str(ROOT / "04-requirements"))
from build_review_table import latest_stm_results, join_stm_topics, make_review_table
import pandas as pd
sent_runs = sorted((ROOT / "02-sentences/data/output/youtube_sent").glob("youtube_sent_*"))
sentence_csv = sent_runs[-1] / "corpus_sentencas.csv"
stm_csv = latest_stm_results(ROOT / "03-topic-modeling/data/output/youtube_doc/stm")
linked = join_stm_topics(pd.read_csv(sentence_csv), pd.read_csv(stm_csv) if stm_csv else None)
review = make_review_table(linked)
out = ROOT / "04-requirements/data/output" / sentence_csv.parent.name / "requirement_review.csv"
out.parent.mkdir(parents=True, exist_ok=True)
review.to_csv(out, index=False)
print(f"{len(review)} review rows -> {out}; stm_linked={stm_csv is not None}")
```
