# 02 — Sentenças YouTube

Gera o corpus de sentenças com contexto e proveniência. `legenda` recebe restauração de pontuação; `whisper` é segmentado diretamente.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().resolve().parents[1]
sys.path.insert(0, str(ROOT / "02-sentences"))
from build_sentences import resolve_latest_input, build_sentence_corpus, _punctuation_model
import pandas as pd
from datetime import datetime
source = resolve_latest_input(ROOT / "01-preprocessing/data/output/youtube") / "corpus_limpo.csv"
docs = pd.read_csv(source).drop(columns=["message_raw"], errors="ignore")
model = _punctuation_model() if docs.source.eq("legenda").any() else None
sentences = build_sentence_corpus(docs, punctuation_model=model)
out = ROOT / "02-sentences/data/output/youtube_sent" / f"youtube_sent_{datetime.now():%Y%m%d_%H%M%S}" / "corpus_sentencas.csv"
out.parent.mkdir(parents=True, exist_ok=True)
sentences.to_csv(out, index=False)
print(f"{len(docs)} docs -> {len(sentences)} sentences -> {out}")
```
